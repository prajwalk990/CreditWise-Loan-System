```python
import seaborn as sns
import matplotlib.pyplot as plt 
```

```python
sns.set_theme()
```

```python
sns.get_dataset_names()
```

```python
tips = sns.load_dataset("tips")
```

```python
sns.set_theme(style="darkgrid")
sns.relplot(
  data=tips ,
    x = "total_bill",
    y="tip",
    hue="sex",
    size='size',
    style="smoker"
)
```

```python
x_val = [1,2,3,4,5]
y_val = [i**2 for i in x_val ]
sns.relplot(
  x=x_val ,
  y=y_val ,
  kind="line"
)
```

```python

tips = sns.load_dataset("tips")
sns.lineplot(
    data = tips ,
    x="total_bill",
    y="tip",
   # hue="time"
)
```

```python
flights = sns.load_dataset("flights")
sns.catplot(
    data = flights,
    x = "month",
    y = "passengers",
    kind="violin"
)
plt.suptitle("FLIGHTS DATA ")
plt.tight_layout()
```

```python
flights.head()
```

```python
flights_pivot = flights.pivot(index="month",columns="year",values="passengers")
```

```python
sns.heatmap(
  data = flights_pivot ,
   cmap="coolwarm",
    annot=True ,
    fmt="d"
)
plt.xticks(rotation=0 ,label="Year")
plt.yticks(rotation=0 ,label="Month")
plt.tight_layout()
plt.show()
```
