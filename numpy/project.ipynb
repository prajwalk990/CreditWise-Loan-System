```python
import json 
```

```python
def load_data(filename):
    with open(filename , 'r') as f :
      data = json.load(f);
    return data 
```

```python
  assign = {
      "one":'1',
      "two":'2',
      "three":'3',
      "four":'4',
      "five":'5',
  }
    
    
      
      
```

```python
def clean_data(data):
    cleaned_data = [];
    check = set();
    for user in data :
         rate = str(user["rating"]);
         user["rating"]= rate.strip();
         user["rating"] = assign.get(rate) if assign.get(rate) else rate ;
         rate = user["rating"];
         user["rating"] = int(round(float(rate)));
         user["age"]= None if user.get("age")==None else user.get("age");
        
       
         val = True if user["name"] in check else False ;
         if(val==True) :
             continue ;
         else :
          check.add(user["name"]);
          cleaned_data.append(user);

    return cleaned_data ;
        
        
    
  
```

```python
 clear_data = clean_data(data)
 clear_data
```

```python
def avg_rating(data) :
    rate = 0 ;
    users = len(data);
    for user in data : 
        rate +=user["rating"];
    return ( rate / users) ;
    
```

```python
avg = avg_rating(clear_data);
print(avg);
```

```python
def  outcome(data) :
    rate = 0 ;
    users = len(data);
    for user in data : 
        if(user["rating"]<3):
            rate +=1;
    return ( rate /users)*100 ;
```
